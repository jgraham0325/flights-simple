# Purpose:
Used for running pytest tests from a notebook within the Databricks workspace.
Can run it locally from VS Code but better to use the VS Code testing extension, or run pytest from command line.

**Important note**: This notebook currently uses an environments yaml (./base_environment_tests.yaml) to set python dependencies, including client version. This can only be viewed by opening this notebook in text format, or from the Databricks UI. Arguably this is overkill for the test notebook, it's just as quick to do a "%pip install..." from the notebook, or set as dependencies directly in the ipynb notebook (also only visible from text editor). 

```python
import sys

import pytest


def run_pytest(pytest_path):
    # Skip writing pyc files on a readonly filesystem.
    sys.dont_write_bytecode = True

    retcode = pytest.main([pytest_path, "-p", "no:cacheprovider", "-p", "no:warnings"])

    # Fail the cell execution if we have any test failures.
    assert retcode == 0, "The pytest invocation failed. See the log above for details."
```

```python
## This path change is needed to make this run with Databricks Workflows (at least when kicking off from VS Code Extension).
## It is not needed to run with databricks-connect or if your test notebook is in the project directory.
import os

current_dir = os.getcwd()
if current_dir.split("/")[-1] == "tests":
    root_dir = os.path.dirname(current_dir)
    print("Root dir:", root_dir)
    os.chdir(root_dir)
    sys.path.append(f"{root_dir}/src")
```

```python
run_pytest("tests/unit_transforms")
```

```python
run_pytest("tests/unit_utils")
```
